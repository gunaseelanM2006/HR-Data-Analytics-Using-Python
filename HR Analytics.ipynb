```python
# ==========================================================
# Chart 1 : Employee Attrition Distribution
# ==========================================================

plt.figure(figsize=(7,5))

ax = sns.countplot(x='Attrition', data=df)

plt.title("Employee Attrition Distribution", fontsize=15, fontweight='bold')
plt.xlabel("Attrition")
plt.ylabel("Number of Employees")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

### Observation

- Most employees stayed with the company.
- A smaller percentage of employees left the organization.
- This indicates that employee attrition is relatively low compared to employee retention.

```python
plt.figure(figsize=(7,5))

ax = sns.countplot(x='Gender', data=df)

plt.title("Gender Distribution", fontsize=15, fontweight='bold')
plt.xlabel("Gender")
plt.ylabel("Employee Count")

for container in ax.containers:
    ax.bar_label(container)

plt.show()
```

# 📌 Task 4: Key Performance Indicator (KPI) Analysis

## 🎯 Objective

The objective of this section is to calculate the Key Performance Indicators (KPIs) that provide a high-level overview of the organization's workforce. These KPIs help management understand employee retention, workforce demographics, compensation, and overall organizational performance.||

```python
# ==========================================================
# TASK 4 : KEY PERFORMANCE INDICATORS (KPI)
# ==========================================================

print("="*80)
print("                 HR ANALYTICS - KPI ANALYSIS")
print("="*80)

# -----------------------------
# Basic KPIs
# -----------------------------

total_employees = len(df)

employees_left = df[df['Attrition'] == 'Yes'].shape[0]

active_employees = df[df['Attrition'] == 'No'].shape[0]

attrition_rate = (employees_left / total_employees) * 100

average_age = df['Age'].mean()

average_income = df['MonthlyIncome'].mean()

average_years = df['YearsAtCompany'].mean()

average_job_satisfaction = df['JobSatisfaction'].mean()

highest_income = df['MonthlyIncome'].max()

lowest_income = df['MonthlyIncome'].min()

most_common_job = df['JobRole'].mode()[0]

largest_department = df['Department'].mode()[0]

highest_attrition_department = (
    df[df['Attrition']=="Yes"]
    ['Department']
    .value_counts()
    .idxmax()
)

# -----------------------------
# Display KPIs
# -----------------------------

print(f"📌 Total Employees                 : {total_employees}")

print(f"📌 Active Employees                : {active_employees}")

print(f"📌 Employees Left                  : {employees_left}")

print(f"📌 Attrition Rate                  : {attrition_rate:.2f}%")

print(f"📌 Average Age                     : {average_age:.2f} Years")

print(f"📌 Average Monthly Income          : ${average_income:.2f}")

print(f"📌 Highest Monthly Income          : ${highest_income}")

print(f"📌 Lowest Monthly Income           : ${lowest_income}")

print(f"📌 Average Years at Company        : {average_years:.2f}")

print(f"📌 Average Job Satisfaction        : {average_job_satisfaction:.2f}")

print(f"📌 Largest Department              : {largest_department}")

print(f"📌 Most Common Job Role            : {most_common_job}")

print(f"📌 Department with Highest Attrition : {highest_attrition_department}")

# -----------------------------
# Business Insights
# -----------------------------

print("\n" + "="*80)
print("                 BUSINESS INSIGHTS")
print("="*80)

print(f"""
1. The organization consists of {total_employees} employees.

2. The overall attrition rate is {attrition_rate:.2f}%.

3. A total of {employees_left} employees have left the organization,
   while {active_employees} employees are currently active.

4. The average employee age is {average_age:.2f} years.

5. Employees earn an average monthly income of ${average_income:.2f}.

6. The average employee tenure is {average_years:.2f} years.

7. The largest workforce belongs to the '{largest_department}' department.

8. The most common job role is '{most_common_job}'.

9. The '{highest_attrition_department}' department experiences the highest employee attrition.

10. These KPIs provide valuable insights into workforce demographics,
    employee retention, compensation, and organizational performance.
""")

print("="*80)
print("TASK 4 COMPLETED SUCCESSFULLY")
print("="*80)
```

# 📌 Task 5: Advanced HR Analytics

## 🎯 Objective

This section performs advanced HR analytics to identify workforce trends and employee attrition patterns across different demographic and organizational factors.

### Analysis Covered

- Attrition by Department
- Attrition by Job Role
- Attrition by Gender
- Attrition by Overtime
- Attrition by Age Group
- Monthly Income vs Attrition

```python
# ==========================================================
# TASK 5 : ADVANCED HR ANALYTICS
# ==========================================================

fig, axes = plt.subplots(3, 2, figsize=(18, 18))

# ----------------------------------------------------------
# 1. Attrition by Department
# ----------------------------------------------------------
sns.countplot(
    data=df,
    x='Department',
    hue='Attrition',
    ax=axes[0,0]
)
axes[0,0].set_title("Attrition by Department", fontsize=14, fontweight='bold')
axes[0,0].tick_params(axis='x', rotation=20)

# ----------------------------------------------------------
# 2. Attrition by Job Role
# ----------------------------------------------------------
sns.countplot(
    data=df,
    y='JobRole',
    hue='Attrition',
    order=df['JobRole'].value_counts().index,
    ax=axes[0,1]
)
axes[0,1].set_title("Attrition by Job Role", fontsize=14, fontweight='bold')

# ----------------------------------------------------------
# 3. Attrition by Gender
# ----------------------------------------------------------
sns.countplot(
    data=df,
    x='Gender',
    hue='Attrition',
    ax=axes[1,0]
)
axes[1,0].set_title("Attrition by Gender", fontsize=14, fontweight='bold')

# ----------------------------------------------------------
# 4. Overtime vs Attrition
# ----------------------------------------------------------
sns.countplot(
    data=df,
    x='OverTime',
    hue='Attrition',
    ax=axes[1,1]
)
axes[1,1].set_title("OverTime vs Attrition", fontsize=14, fontweight='bold')

# ----------------------------------------------------------
# 5. Age vs Attrition
# ----------------------------------------------------------
sns.boxplot(
    data=df,
    x='Attrition',
    y='Age',
    ax=axes[2,0]
)
axes[2,0].set_title("Age vs Attrition", fontsize=14, fontweight='bold')

# ----------------------------------------------------------
# 6. Monthly Income vs Attrition
# ----------------------------------------------------------
sns.boxplot(
    data=df,
    x='Attrition',
    y='MonthlyIncome',
    ax=axes[2,1]
)
axes[2,1].set_title("Monthly Income vs Attrition", fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()


print("="*80)
print("               BUSINESS INSIGHTS")
print("="*80)

print("""
• Research & Development has the highest number of employees, which also results in the highest number of attrition cases.

• Sales Executive is one of the most common job roles and should be monitored for employee retention.

• Employees working overtime appear more likely to leave the organization.

• Salary differences may influence employee attrition.

• Employee age and job role can be important indicators of workforce stability.

• HR management can use these findings to improve employee retention strategies.

""")

print("="*80)
print("TASK 5 COMPLETED SUCCESSFULLY")
print("="*80)
```
